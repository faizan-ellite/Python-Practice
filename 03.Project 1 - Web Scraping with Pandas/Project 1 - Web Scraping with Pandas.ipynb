```python
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
# creating list of leagues
leagues = ['E0','E1','E2', 'E3','EC']

frames = []
# looping through leagues, read multiple csv and append it into a list
for league in leagues:
    df = pd.read_csv(root + "2526" + "/"  + league + ".csv")
    frames.append(df)
```

```python
len(frames)
```

```python
# Merge all DataFrames into one DataFrame
Faizan = pd.concat(frames, ignore_index=True)
```

```python
Faizan
```

```python
# Save to CSV
Faizan.to_csv("2025-26 data.xlsx", index=False)
```

```python
leagues = ['E0','E2', 'E3']

frames = []

# loop through range of seasons and insert new column
for league in leagues:
    for season in range(15,21): # range of seasons
        df = pd.read_csv(root + str(season) + str(season+1) + "/" + league + ".csv")
#         df['season'] = season
        
        frames.append(df)
```

```python
len(frames)
```

```python
# Merge all DataFrames into one DataFrame
Faizan = pd.concat(frames, ignore_index=True)
```

```python
leagues = ['E0','E1','E2', 'E3','EC']

frames = []

# loop through range of seasons and insert new column
for league in leagues:
    for season in range(15,26): # range of seasons
        df = pd.read_csv(root + str(season) + str(season+1) + "/" + league + ".csv", encoding='latin1')
        frames.append(df)
```

```python
len(frames)
```

```python
Faizan = pd.concat(frames, ignore_index=True)
```
